message prompts
Message prompts = structured conversation input you send to a chat model

Instead of giving one plain string, you send a list of messages, each with a role.

📦 Basic Structure
[
  {"role": "system", "content": "..."},
  {"role": "user", "content": "..."},
  {"role": "assistant", "content": "..."}
]
Message prompts are structured inputs (list of messages) sent to a chat model instead of a single string.
Each message has a role like system, user, or assistant.
System sets rules/behavior, user asks the question, assistant gives responses.
They help maintain conversation context (chat history) across turns.
Essential for features like tool calling, memory, and multi-step reasoning.

```python
from langchain.messages import SystemMessage,HumanMessage,AIMessage
messages=[
    SystemMessage("You are a poetry expert"),
    HumanMessage("write a poem on ai")

]
response=model.invoke(messages)
response
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Hi, who are you?"),
    AIMessage(content="I am an AI assistant."),
    HumanMessage(content="What's the weather in Boston?")
]
response=model.invoke(messages);
response.content
```

```python
from langchain_core.messages import AIMessage, ToolMessage

# Step 1: AI generates a tool call
ai_message = AIMessage(
    content="",
    tool_calls=[
        {
            "id": "call_1",
            "name": "get_weather",
            "args": {"location": "Boston"}
        }
    ]
)

print("AI Message:", ai_message.tool_calls)

# Step 2: Tool executes (mock result)
tool_result = "It is sunny in Boston"

# Step 3: Send tool result back to model
tool_message = ToolMessage(
    content=tool_result,
    tool_call_id="call_1"
)

print("Tool Message:", tool_message)
messages=[HumanMessage('what weathe in boston'),
          ai_message,
          tool_message]
```
