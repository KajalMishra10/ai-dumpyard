streaming and batch

Streaming and batch are two ways an AI model returns responses. In streaming, the output is sent gradually, token by token, giving a real-time typing effect and improving user experience in chat applications. In batch processing, the model generates the entire response first and then returns it all at once, which is simpler to handle and commonly used in APIs and backend tasks. In short, streaming feels faster and interactive, while batch is more straightforward and consistent.

```python
for chunk in model.stream("why do parrots have colorful feathers"):
    print(chunk.text, end="|", flush=True)
```

Batch

```python
responses=model.batch(["why do parrots have colorful feathers","why do airplanes fly"], config={
    'max_concurrency':5
})
for response  in responses:
    print(response)
```
