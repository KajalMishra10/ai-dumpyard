```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model=init_chat_model("groq:llama-3.3-70b-versatile")
response=model.invoke("why do parrot talks")
response
```

```python
from langchain.tools import tool
from langchain_core.messages import HumanMessage

@tool('get_weather')
def get_weather(location:str)->str:
    """ get the weather at a location"""
    return f"it is sunny at {location}"

model_with_tools=model.bind_tools([get_weather])
```

```python
messages = [{"role": "user", "content": "what's the weather in boston"}]
ai_msg =model_with_tools.invoke(messages)
```

```python
print("AI MSG:", ai_msg)
print("TOOL CALLS:", ai_msg.tool_calls)
```

```python
messages.append(ai_msg)
```

```python
if ai_msg.tool_calls:
    for tool_call in ai_msg.tool_calls:
        print("RUNNING TOOL:", tool_call)
        result = get_weather.invoke(tool_call["args"])
        messages.append({
            "role": "tool",
            "content": result,
            "tool_call_id": tool_call["id"],
        })
else:
    print("No tool call happened")

final_response = model_with_tools.invoke(messages)
print(final_response.content)
```

tool_call_id pehle thoda unnecessary lagta hai, lekin jab ek se zyada tools ya multiple tool calls aati hain tab bahut important ho jata hai.

Example 1: Ek hi tool call

LLM return karta hai:

AIMessage(
    content="",
    tool_calls=[
        {
            "id": "call_abc123",
            "name": "get_weather",
            "args": {"location": "Boston"}
        }
    ]
)

Tool run hua:

it is sunny at Boston

Tool result bhejte waqt:

{
    "role": "tool",
    "content": "it is sunny at Boston",
    "tool_call_id": "call_abc123"
}

Yahan ID se model ko pata chalta hai:

Yeh result meri call_abc123 wali request ka answer hai.

Example 2: Multiple tool calls

User:

Weather in Boston and Delhi?

LLM decide kar sakta hai:

tool_calls = [
    {
        "id": "call_1",
        "name": "get_weather",
        "args": {"location": "Boston"}
    },
    {
        "id": "call_2",
        "name": "get_weather",
        "args": {"location": "Delhi"}
    }
]

Ab tools run hue:

Boston -> Sunny
Delhi  -> Rainy

History mein:

{
    "role": "tool",
    "content": "Sunny",
    "tool_call_id": "call_1"
}

{
    "role": "tool",
    "content": "Rainy",
    "tool_call_id": "call_2"
}

Ab LLM easily map kar sakta hai:

call_1 -> Boston -> Sunny
call_2 -> Delhi  -> Rainy
Agar ID na ho

Socho:

[
    {"role":"tool","content":"Sunny"},
    {"role":"tool","content":"Rainy"}
]

Ab model ko kaise pata chalega:

Sunny Boston ka hai?
Ya Delhi ka?
Rainy kis city ka hai?

Confusion ho sakti hai.

Internally Chat History

 HumanMessage("Weather in Boston and Delhi"),
