```python
from langchain_community.document_loaders import TextLoader

loader=TextLoader("speech.txt",encoding="utf-8")

documents=loader.load();
documents
```

```python
from langchain_text_splitters import CharacterTextSplitter

splitter=CharacterTextSplitter(
    separator='.',
    chunk_size=50,
    chunk_overlap=10
)

chunks=splitter.split_text(documents[0].page_content)

for i, chunk in enumerate(chunks):
    print(f"chunk {i+1}:{chunk}")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter=RecursiveCharacterTextSplitter(chunk_size=60, chunk_overlap=15)
response=splitter.split_text(documents[0].page_content)
response
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["HF_TOKEN"]=os.getenv("HF_TOKEN")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(model="sentence-transformers/all-MiniLM-L6-v2")

embeddings.embed_query(response[0])
```

```python
from langchain_ollama import OllamaEmbeddings
embeddings=OllamaEmbeddings(model="gemma:2b")
r1=embeddings.embed_documents(response)
r1
```

```python
from langchain_chroma import Chroma
vectorStore=Chroma.from_texts(response[0], embeddings)
```

```python
r=vectorStore.similarity_search("what is tinder")
r
```
