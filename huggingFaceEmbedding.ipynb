```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
text="LangGraph is a knowledge graph that connects concepts, entities, and relationships in a structured way. It provides a visual representation of information and allows users to explore and understand complex relationships between different pieces of data. LangGraph can be used for various applications such as data analysis, information retrieval, and knowledge discovery."
vector = embeddings.embed_query(text)
print(vector)
```
