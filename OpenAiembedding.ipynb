```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """LangChain is a powerful framework for building AI applications.
It provides document loaders, text splitters, embeddings, and vector stores.
You can use it to build chatbots, Q&A systems, and AI agents.
It supports OpenAI, Anthropic, Google, and many other LLM providers."""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

chunks = splitter.split_text(text)
```

```python
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
vectorstore = Chroma.from_texts(chunks, embeddings)
```

```python
query="It provides document loaders, "
results = vectorstore.similarity_search(query)
print(results)
```
