```python
import os
from dotenv import load_dotenv
load_dotenv()

import openai

#openai.api_key = os.getenv("HF_TOKEN")
groq_api_key = os.getenv("GROQ_API_KEY")
```

from langchain_groq import ChatGroq
chat_groq = ChatGroq(model="llama-3.1-8b-instant", groq_api_key=groq_api_key)
#response = chat_groq("What is the capital of France?")

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
messages = [
    SystemMessage(content="Translate the following English text to Hindi."),
    HumanMessage(content="What is the capital of France?"),
]

result=chat_groq.invoke(messages)
```

```python
from langchain_core.output_parsers import StrOutputParser
output_parser = StrOutputParser()
output_parser.invoke(result)
```

```python
chain =chat_groq | output_parser
chain.invoke(messages)
```

```python
from langchain_core.prompts import ChatPromptTemplate
generic_messages="translate the following English text to {language}"
prompt = ChatPromptTemplate.from_messages([
    ("system", generic_messages),
    ("human", "{text}")
])
```

```python
result=prompt.invoke({"language":"Hindi","text":"What is the capital of France?"})
```

```python
result.to_messages()
```

```python
chain=prompt | chat_groq | output_parser
chain.invoke({"language":"Hindi","text":"What is the capital of France?"})
```
